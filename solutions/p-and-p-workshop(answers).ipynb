# Making Choices with Conditionals
https://swcarpentry.github.io/python-novice-inflammation/07-cond/index.html

When we start working with a lot of data, we might want to start treating different groups of data different from others. Let's see how in the next cell.

````python
number = 15
if number > 0:
    print(number, "is positive!")
````

This is helpful for positive numbers, but I also want to account for negative numbers.


````python
number = -15 
if number > 0:
    print(number, "is positive!")
else:
    print(number, "is negative!")
````

But wait, what if I input zero, I don't think that's negative or positive.

````python
number = 0
if number > 0:
    print(number, "is positive!")
elif number < 0:
    print(number, "is negative")
else:
    print(number, "is zero!")
````

General Structure of `if statement`:  

    if Boolean Expression:
        statements
    elif Boolean Expression:
        statements
    elif ... (You can have an indefinite number of elif statements)
    else:
        statements

I've picked three of my favorite numbers. Let's use our conditional statements to sort each one: positive, negative, or zero.

````python
number1 = 8
number2 = -12
number3 = 0

if number1 > 0:
    print(number1, "is positive!")
elif number1 < 0:
    print(number1, "is negative!")
else:
    print(number1, "is zero!")

if number2 > 0:
    print(number2, "is positive!")
elif number2 < 0:
    print(number2, "is negative!")
else:
    print(number2, "is zero!")

if number3 > 0:
    print(number3, "is positive!")
elif number3 < 0:
    print(number3, "is negative!")
else:
    print(number3, "is zero!")
````

In the first section, we mentioned **boolean values**.

Boolean values are **True** or **False**.

These are most often calculated using relational operators. Some examples of relational operators are:


```
# Check if two values are equal (==)
5 == 1
False

#Check if two values are NOT equal (!=)
5 != 1
True

# Greater than or equal to (>=)
5 >= 1
True

# Less than or equal to (<=)
5 <= 1
False

# Less than (<)
5 < 5
False

# Greater than (>)
5 > 1
True
```



Conditionals are **if**, **elif**, and **else** statements. If and elif statements run the code inside them only if their boolean value is True, the else statement will be used.


```
if True:
    # Run This Code
elif True:
    # Run this code
else:
    # Everything else was False, run this code.
```

**Note:** All numbers (except 0) and strings on their own are evaluted as True values.

````python
if 1:
    print(True)
````

````python
if 0:
    print(True)
else:
    print(False)
````

````python
if "Anything":
    print(True)
````

# Final Activity: Who Ate the Most Pizza?

We ordered pizza after the workshop for five friends. We wrote down how many slices each person ate, how much they spent, and how happy they were afterwards (on a scale of 1 to 15):

| name | slices_eaten | dollars_spent | happiness_rating |
|------|--------------|---------------|------------------|
| Theo | 14 | 15 | 14 |
| Barack | 6 | 9 | 9 |
| Taylor | 8 | 8 | 10 |
| Jelly | 4 | 2 | 12 |
| Kool | 0 | 1 | 15 |

Yes, Kool is the Kool-Aid Man. He's a pitcher. He can't eat pizza. He paid a dollar for a napkin and had a wonderful time.

Each list keeps our five friends in the same order: `[0]` is Theo, `[1]` is Barack, `[2]` is Taylor, `[3]` is Jelly, `[4]` is Kool. (Python starts counting at 0!)

````python
#our pizza data: five friends, four lists, same order in each
names = ["Theo", "Barack", "Taylor", "Jelly", "Kool"]
slices_eaten = [14, 6, 8, 4, 0]
dollars_spent = [15, 9, 8, 2, 1]
happiness_rating = [14, 9, 10, 12, 15]
````

### **Question 1:** Who ate the most pizza?

**Before we run any code — vote!** Raise a hand: Theo? Barack? Taylor? Jelly? Kool?

*(If the room is quiet: "I'll guess Theo — let's find out.")*

````python
#max() is a built-in: it hands us the biggest value in a list
most_slices = max(slices_eaten)
most_slices
````

````python
if most_slices == slices_eaten[0]:
    print(names[0], "ate the most pizza!")
elif most_slices == slices_eaten[1]:
    print(names[1], "ate the most pizza!")
elif most_slices == slices_eaten[2]:
    print(names[2], "ate the most pizza!")
elif most_slices == slices_eaten[3]:
    print(names[3], "ate the most pizza!")
else:
    print(names[4], "ate the most pizza!")
````

### **Exercise:** Who got the best value?

Which friend got the most slices per dollar? **Vote first!**

*(If the room is quiet: "Theo ate the most, so I'll guess Theo again — let's find out.")*

*(Hint: slices per dollar = slices_eaten / dollars_spent)*

*(Hint: [0] is Theo, [1] is Barack, [2] is Taylor, [3] is Jelly, [4] is Kool)*

````python
#slices per dollar for each friend
theo_value = slices_eaten[0] / dollars_spent[0]
barack_value = slices_eaten[1] / dollars_spent[1]
taylor_value = slices_eaten[2] / dollars_spent[2]
jelly_value = slices_eaten[3] / dollars_spent[3]
kool_value = slices_eaten[4] / dollars_spent[4]

best_value = max(theo_value, barack_value, taylor_value, jelly_value, kool_value)

if best_value == theo_value:
    print("Theo got the best value!")
elif best_value == barack_value:
    print("Barack got the best value!")
elif best_value == taylor_value:
    print("Taylor got the best value!")
elif best_value == jelly_value:
    print("Jelly got the best value!")
else:
    print("Kool got the best value!")
````

## But what if it wasn't 5 friends?

This was easy with 5 rows you can see at a glance. But what if we had pizza data for **20 million people**? You couldn't read that list. You couldn't hold it in your head. You practically NEED computers.

And here's the best part: **the exact same code still works.**

````python
#this cell just cooks the pizza data - you don't need to understand it yet!
#(it makes a list of 5000 fake pizza eaters, each ate between 0 and 15 slices)
slices_big = [(i * 7919 + 3) % 16 for i in range(5000)]

print(len(slices_big), "people at the pizza party")
````

````python
#the exact same question, 5000 rows, instantly
most_slices_big = max(slices_big)

print("The most slices eaten by anyone:", most_slices_big)
print("Out of", len(slices_big), "people!")
print("Total slices eaten:", sum(slices_big))
````

### **Finished early? Bonus question!** The Happiest Dollar

Finding the best *value* was slices per dollar — but what if we cared about **joy per dollar**? Who got the most happiness for each dollar they spent?

**Vote first!** Theo? Barack? Taylor? Jelly? Kool?

*(If the room is quiet: "Kool barely spent anything, so I'll guess Kool — let's find out.")*

*(Hint: happiness per dollar = happiness_rating / dollars_spent)*

*(Hint: [0] is Theo, [1] is Barack, [2] is Taylor, [3] is Jelly, [4] is Kool)*

````python
#happiness per dollar for each friend
theo_happiness = happiness_rating[0] / dollars_spent[0]
barack_happiness = happiness_rating[1] / dollars_spent[1]
taylor_happiness = happiness_rating[2] / dollars_spent[2]
jelly_happiness = happiness_rating[3] / dollars_spent[3]
kool_happiness = happiness_rating[4] / dollars_spent[4]

happiest_dollar = max(theo_happiness, barack_happiness, taylor_happiness, jelly_happiness, kool_happiness)

if happiest_dollar == theo_happiness:
    print("Theo got the happiest dollar!")
elif happiest_dollar == barack_happiness:
    print("Barack got the happiest dollar!")
elif happiest_dollar == taylor_happiness:
    print("Taylor got the happiest dollar!")
elif happiest_dollar == jelly_happiness:
    print("Jelly got the happiest dollar!")
else:
    print("Kool got the happiest dollar!")

#expected output: Kool got the happiest dollar!
````
